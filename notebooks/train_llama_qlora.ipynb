```python
import os
import torch
import wandb
from huggingface_hub import login

# 1. Verify CUDA GPU Allocation
print(f"CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU Device: {torch.cuda.get_device_name(0)}")
else:
    raise SystemError("GPU not attached. Check your Colab/Kaggle runtime settings.")

# 2. Agnostic Secret Fetching
try:
    # Attempt Google Colab Secret Fetching
    from google.colab import userdata
    hf_token = userdata.get('HF_TOKEN')
    wandb_key = userdata.get('WANDB_API_KEY')
    env_name = "Google Colab"
except ImportError:
    # Fallback to Kaggle Secret Fetching
    from kaggle_secrets import UserSecretsClient
    secrets = UserSecretsClient()
    hf_token = secrets.get_secret('HF_TOKEN')
    wandb_key = secrets.get_secret('WANDB_API_KEY')
    env_name = "Kaggle"

print(f"Successfully loaded secrets for: {env_name}")

# 3. Authenticate
login(token=hf_token)
wandb.login(key=wandb_key)
```

```python
# Configure storage paths dynamically
if env_name == "Google Colab":
    from google.colab import drive
    drive.mount('/content/drive')
    CHECKPOINT_DIR = "/content/drive/MyDrive/Capstone_Models/"
else:
    # Kaggle output directory
    CHECKPOINT_DIR = "/kaggle/working/Capstone_Models/"

os.makedirs(CHECKPOINT_DIR, exist_ok=True)
print(f"Model checkpoints and quantized .gguf files will save to: {CHECKPOINT_DIR}")
```
